```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("/content/heart.csv")
df.info()
# There is no missing/null values
```

```python
df = df.drop_duplicates()
df.info()
# There is no duplicate rows
```

```python
df = df[(df['RestingBP'] != 0) & (df['Cholesterol'] != 0)]
df
```

```python
df
```

```python
g = sns.lmplot(data=df, x='Age', y='Cholesterol', scatter_kws={"s": 10})
g.fig.suptitle('Relationship Between Cholesterol Levels and Age', y=1.03) # Add title to the figure
plt.show()
print('\n The r-value is',df[['Age','Cholesterol']].corr().loc['Age', 'Cholesterol'], '\n This is a very weak positive correlation')
```
